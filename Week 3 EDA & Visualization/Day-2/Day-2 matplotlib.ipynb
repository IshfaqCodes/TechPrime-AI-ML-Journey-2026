```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from numpy.random import randint
```

## Distribution and Advanced Plots
### Histogram (hist())

```python
numbers=[23,34,45,77,88,12,1,90,78,56]
plt.hist(numbers,bins=13,density=True)
plt.show 
```

###  Scatter Plot

```python
Age=randint(1,10,20)
Marks=randint(1,10,20)

scater=plt.scatter(Age,Marks)
plt.show
```

### Bubble Chart

```python
x = [1,2,3,4,5] 
y = [10,20,15,25,30] 
sizes = [100,200,300,400,500] 
plt.scatter(x, y, s=sizes) 
plt.show() 
```

### Pie Chart 



```python
labels = ['A','B','C','D','E'] 
sizes = [30,40,30,20,12] 
plt.pie(sizes, labels=labels,autopct='%1.1f%%') 
plt.show() 
```

### Box plot

```python
marks=[23,34,45,77,88,12,1,90,78,56,200]
plt.boxplot(marks)
plt.show 
```

```python
data = [marks, [60,65,70,75,80,-12]] 
plt.boxplot(data) 
plt.show() 
```

### Figure and Axes Concept

```python
plt.figure(figsize=(8,5))
plt.plot([1,2,3], [10,20,30])
plt.show()
```

### Subplots 

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.plot([1,2,3], [10,20,30])
plt.subplot(1,2,2)
plt.plot([1,2,3], [5,15,25])
plt.show()
```

###  Annotations

```python
x = [1,2,3]
y = [10,20,30]
plt.plot(x, y)
plt.annotate('Peak', xy=(3,30), xytext=(2,25), arrowprops=dict())
plt.show()
```

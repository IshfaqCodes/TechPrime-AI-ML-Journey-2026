```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from numpy.random import randint
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
iris = sns.load_dataset("iris")
tips
```

## 1.Relational Plot
### a)Sacatter plot

```python
# Scatter plot>- axis level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#Adding hue
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
# Adding style
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time')
```

```python
# Using size 
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# Relplot >- square shape 
sns.relplot(data=tips,x='total_bill',y='tip')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
sns.scatterplot(x="sepal_length", y="sepal_width", data=iris) 
plt.show()
```

### b)line plot

```python
flights = sns.load_dataset("flights") 
sns.lineplot(x="year", y="passengers", data=flights) 
plt.show() 
```

## 2.Distribution Plots

### a)Histogram using histplot
use hota hai jab hume ek numerical column ki distribution dekhni ho

```python
titanic = sns.load_dataset("titanic") 
sns.histplot(data=titanic, x="age", bins=20) # hamaree pass zada log kis age k theh it show the distribution
plt.show()
```

### b)Kde Plot

```python

sns.kdeplot(data=iris, x="sepal_length") 
plt.show() 
```

```python
# Controlling KDE Curve
sns.kdeplot(data=iris, x="sepal_length", bw_adjust=0.5) 
plt.show() 
```

```python
#combining  hist and kde plot
sns.histplot(data=iris, x="petal_length", kde=True) 
plt.show() 
```

### c)Rug Plot

```python
sns.rugplot(data=iris, x="sepal_length") 
plt.show() 
```

## 3.Categorical Plots

### a)countplot()

```python
sns.countplot(x="sex", data=titanic) 
plt.show() 
```

### b) Barplot

```python
sns.barplot(x="class", y="fare", data=titanic) # kis class main logo ne zada kraya dyaa 
plt.show()
```

### c)Boxplot

```python
sns.boxplot(x="class", y="age", data=titanic) 
plt.show() 
```

### d) Violinplot

```python
tips = sns.load_dataset("tips") 
sns.violinplot(x="day", y="total_bill", data=tips) 
plt.show()
```

### e) Stripplot

```python
sns.stripplot(x="day", y="total_bill", data=tips) 
plt.show() 
```

### f) Swarmplot

```python
sns.swarmplot(x="day", y="total_bill", data=tips) 
plt.show() 
```

### Combining Plots

```python

sns.boxplot(x="day", y="total_bill",hue='sex',palette='dark:red',data=tips) 
sns.stripplot(x="day", y="total_bill",hue='sex' ,data=tips) 
plt.show() 
```

```python
import pandas as pd 
corr = titanic.corr(numeric_only=True) #Correlation matrix batata hai ke variables ke darmiyan relation kitna strong hai.
print(corr) 
sns.heatmap(corr) 
plt.show()
```

```python
sns.heatmap(corr, annot=True) 
plt.show()
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm") 
plt.show() 
```

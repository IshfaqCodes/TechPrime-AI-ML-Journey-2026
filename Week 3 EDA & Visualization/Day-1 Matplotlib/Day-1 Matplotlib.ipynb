# Importing Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from numpy.random import randint
```

```python
x=np.linspace(1,10,20)
y=randint(1,50,20)
print(x)
print(y)
```

###  Basic Line Plot using plot

```python
plt.plot(y)
plt.title("Y")
plt.grid()
plt.show()
```

```python
plt.plot(x)
plt.title("Y")
plt.grid()
plt.show()
```

```python
plt.plot(x,y)
plt.plot(y,x)
```

```python
Time=np.array([3,4,5,6,7])  #Matplotlib lists aur arrays dono ke sath kaam karta hai to hum array k zaaryee krengee 
sales=np.array([10,23,44,55,99])
plt.plot(Time,sales)
plt.show()
```

### Adding Title and Labels

```python
Time=[3,4,5,6,7]
sales=[10,23,44,55,99]
plt.plot(Time,sales)
plt.title("Sales vs Time")
plt.xlabel("Time")
plt.ylabel("Sales")
plt.grid()  #Grid background lines hoti hain jo graph ko read karna easy banati hain
plt.show()
```

### Customization Color, Linewidth, Linestyle
- color line ka color change karta hai.
- linewidth line ki thickness control karta hai.
- linestyle line ka pattern change karta hai

```python
Time=[3,4,5,6,7]
sales=[10,23,44,55,99]
plt.plot(Time,sales,color="red",linewidth=2,linestyle="--",marker=0) #Markers har data point ko highlight karte hain
plt.title("Sales vs Time")
plt.xlabel("Time")
plt.ylabel("Sales")
plt.grid()  #Grid background lines hoti hain jo graph ko read karna easy banati hain
plt.show()
```

```python
sales2=np.array([20,88,99,77,88]) #new array sales2 for second barr
plt.plot(Time, sales, label='Sales 2023') 
plt.plot(Time, sales2, label='Sales 2024') 
plt.grid()  #Grid background lines hoti hain jo graph ko read karna easy banati hain

plt.legend() 
plt.show() 
```

## Working with Real Dataset

```python
months =np.array(['Jan','Feb','Mar','Apr','May']) 
sales =np.array([200,250,300,280,350])
plt.plot(months,sales,color="blue",linewidth=2,linestyle="--",marker=0)
plt.title("Months vs Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

## Plotting using Pandas DataFrame

```python
data = { 
'month': ['Jan','Feb','Mar','Apr'], 
'sales': [100,200,150,300] 
} 
df = pd.DataFrame(data) 
plt.plot(df['month'], df['sales'],color="red",marker=0,linewidth=3,linestyle="--") 
```

## 2. bar() Plot
Bar plot rectangular bars use karke data ko represent karta hai.
Har bar ek category ko represent karti hai

```python
data={
    "subjects":["English","Math","Bio"],
    "marks":[30,50,90]
}
df=pd.DataFrame(data)
plt.bar(df["subjects"],df["marks"],color="red",linewidth=2,linestyle="--",width=0.2)  # width bars ki thickness control karta hai
plt.title("Sunjects vs Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

###  Horizontal Bar Plot (barh())

```python
data={
    "subjects":["English","Math","Bio"],
    "marks":[30,50,90]
}
df=pd.DataFrame(data)
plt.barh(df["subjects"],df["marks"],color="red",linewidth=2,linestyle="--")
plt.title("Sunjects vs Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

###  Stacked Bar Chart

```python
boys=[20,50,40]
girl=[22,11,22]
subjects=["math","english","bio"]
plt.bar(subjects,boys,bottom=girl,label="Boys")
plt.bar(subjects,girl,label="Girls")
plt.legend()
plt.grid()
plt.show()
```

### Adding Labels on Bars

```python
for i in range(len(df)):
    plt.text(i, df["marks"][i], df["marks"][i])

plt.bar(df["subjects"], df["marks"])
plt.show()
```

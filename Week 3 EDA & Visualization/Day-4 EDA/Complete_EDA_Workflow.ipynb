## Phase 2: Data Quality Check

### 2.1 Missing Values

```python
df.isnull().sum()
```

### 2.2 Missing Value Percentage

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': missing_percent.round(2)
}).sort_values('Missing_Percentage', ascending=False)

missing_df[missing_df['Missing_Count'] > 0]
```

### 2.2 Missing Values

**Explanation:** This table shows the percentage of missing values in each column.

**Interpretation:** If a column has a small percentage of missing values, they can usually be handled by removing or imputing the missing values. If a column has a very high percentage of missing values, removing the column may be considered.

### 2.3 Duplicate Rows


```python
duplicate_rows = df.duplicated().sum()
print(f"Total duplicate rows: {duplicate_rows}")
```

### 2.4 Duplicate Patient_ID

```python
duplicate_ids = df['Patient_ID'].duplicated().sum()
print(f"Duplicate Patient_IDs: {duplicate_ids}")

# Duplicate ID rows dekhne ke liye
df[df['Patient_ID'].duplicated(keep=False)].sort_values('Patient_ID')
```

**Interpretation:** `Patient_ID` should be a unique identifier for each patient. If duplicate IDs are found, they may indicate data entry errors or duplicate records.

### 2.5 Unique Values in Each Column

### 2.7 Convert Date Columns to Datetime

```python
df['Diagnosis_Date'] = pd.to_datetime(df['Diagnosis_Date'], errors='coerce')

df['Treatment_Start_Date'] = pd.to_datetime(
    df['Treatment_Start_Date'], errors='coerce'
)

df['Death_Date'] = pd.to_datetime(
    df['Death_Date'], errors='coerce'
)

df[['Diagnosis_Date', 'Treatment_Start_Date', 'Death_Date']].dtypes
```

```python
df.groupby('Hospital').agg(
    Total_Patients=('Patient_ID', 'count'),
    Avg_Age=('Age', 'mean'),
    Avg_BMI=('BMI', 'mean'),
    Avg_Treatment_Delay=('Treatment_Delay', 'mean')
).round(2).sort_values('Total_Patients', ascending=False)
```

## Phase 8: Correlation Analysis


```python
# Correlation Matrix
corr_matrix = df[numerical_cols + ['BMI', 'Treatment_Delay', 'Survival_Days']].corr()
corr_matrix.round(2)

# Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap - Numerical Features')
plt.tight_layout()
plt.show()
```

**Interpretation:** `Weight` and `BMI` are expected to have a strong positive correlation. The relationship between `Treatment_Delay` and `Survival_Days` can also provide useful insights into the importance of treatment timing.

## Phase 9: Time Series Analysis


```python
# Yearly Diagnosis Trend
yearly_trend = df['Diagnosis_Year'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
sns.lineplot(x=yearly_trend.index, y=yearly_trend.values, marker='o', color='darkblue')
plt.title('Yearly Diagnosis Trend')
plt.xlabel('Year')
plt.ylabel('Number of Diagnoses')
plt.show()

# Monthly Diagnosis Trend
month_order = ['January','February','March','April','May','June','July','August','September','October','November','December']
monthly_trend = df['Diagnosis_Month'].value_counts().reindex(month_order)

plt.figure(figsize=(12, 6))
sns.lineplot(x=monthly_trend.index, y=monthly_trend.values, marker='o', color='darkred')
plt.title('Monthly Diagnosis Trend')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Treatment Delay Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['Treatment_Delay'].dropna(), kde=True, color='green')
plt.title('Treatment Delay Distribution (Days)')
plt.show()

# Survival Days Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['Survival_Days'].dropna(), kde=True, color='purple')
plt.title('Survival Days Distribution')
plt.show()
```

## Time-Series Analysis

**Explanation:** Time-series analysis shows diagnosis patterns by year or month. It also helps understand treatment delays and survival patterns.

**Interpretation:** If there is a sudden increase in diagnoses in a particular year, possible reasons such as screening programs or awareness campaigns can be investigated. Longer `Treatment_Delay` may be associated with poorer outcomes.

**Explanation:** The IQR method identifies values that are below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR`.

**Interpretation:** These outliers may be caused by data entry errors, such as an incorrect height value, or they may represent genuine extreme cases. Therefore, they should be checked using domain knowledge before removing them.

## Phase 11: Insights


```python
print("KEY INSIGHTS")
print("=" * 50)

print(f"1. Most common cancer type: {df['Cancer_Type'].mode()[0]}")

most_affected_age_group = df['Age_Group'].value_counts().idxmax()
print(f"2. Most affected age group: {most_affected_age_group}")

print(f"3. Most common treatment: {df['Treatment_Type'].mode()[0]}")

smoking_outcome = pd.crosstab(df['Smoking_Status'], df['Outcome'], normalize='index') * 100
print(f"4. Smoking vs Outcome (%):\n{smoking_outcome.round(1)}")

hospital_performance = df.groupby('Hospital')['Outcome'].apply(
    lambda x: (x == 'Recovered').mean() * 100
).sort_values(ascending=False)
print(f"5. Hospital performance (Recovery %):\n{hospital_performance.round(1)}")

stage_mortality = pd.crosstab(df['Cancer_Stage'], df['Outcome'], normalize='index') * 100
print(f"6. Stage with highest mortality:\n{stage_mortality.round(1)}")
```

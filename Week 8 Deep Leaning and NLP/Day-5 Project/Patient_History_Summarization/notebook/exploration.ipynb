# Patient History Summarization — Data Exploration

```python
import sys
sys.path.append('../src')
import pandas as pd
from preprocess import load_dataset, split_dataset

df = load_dataset('../data/sample_patient_histories.csv')
df.head()
```

```python
# Basic stats: text length distribution
df['history_word_count'] = df['patient_history'].str.split().apply(len)
df['summary_word_count'] = df['reference_summary'].str.split().apply(len)
df[['history_word_count', 'summary_word_count']].describe()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(df['history_word_count'], bins=8)
ax[0].set_title('Patient History Word Count')
ax[1].hist(df['summary_word_count'], bins=8)
ax[1].set_title('Reference Summary Word Count')
plt.tight_layout()
plt.show()
```

```python
# Compression ratio: how much shorter are summaries vs. original text?
df['compression_ratio'] = df['summary_word_count'] / df['history_word_count']
print('Average compression ratio:', df['compression_ratio'].mean().round(2))
```

```python
# Train/val/test split preview
train_df, val_df, test_df = split_dataset(df)
print('Train:', len(train_df), 'Val:', len(val_df), 'Test:', len(test_df))
train_df.head()
```

# Day 3: Regularization + Overfitting + NN Architectures (FNN/CNN/RNN)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

## 3.1 Overfitting Demo (Polynomial Regression)

```python
np.random.seed(0)
X_poly = np.sort(np.random.rand(30))
y_poly = np.sin(2 * np.pi * X_poly) + np.random.randn(30) * 0.1

degrees = [1, 4, 15]
plt.figure(figsize=(12, 4))
for i, d in enumerate(degrees):
    coeffs = np.polyfit(X_poly, y_poly, d)
    poly = np.poly1d(coeffs)
    x_line = np.linspace(0, 1, 200)
    plt.subplot(1, 3, i + 1)
    plt.scatter(X_poly, y_poly)
    plt.plot(x_line, poly(x_line))
    plt.title(f"Degree {d}")
plt.show()
```

## 3.2 L2 Regularization + Dropout with Keras

```python
X_moons, y_moons = make_moons(n_samples=500, noise=0.2, random_state=1)
X_tr, X_te, y_tr, y_te = train_test_split(X_moons, y_moons, test_size=0.2, random_state=1)
scaler = StandardScaler().fit(X_tr)
X_tr, X_te = scaler.transform(X_tr), scaler.transform(X_te)

reg_model = keras.Sequential([
    layers.Dense(32, activation="relu", kernel_regularizer=keras.regularizers.l2(0.01),
                 input_shape=(2,)),
    layers.Dropout(0.3),
    layers.Dense(16, activation="relu", kernel_regularizer=keras.regularizers.l2(0.01)),
    layers.Dropout(0.3),
    layers.Dense(1, activation="sigmoid"),
])
reg_model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
reg_model.fit(X_tr, y_tr, epochs=20, batch_size=16, validation_data=(X_te, y_te), verbose=0)
print("Regularized model test acc:", reg_model.evaluate(X_te, y_te, verbose=0)[1])
```

## 3.3 FNN (Fully-Connected Feedforward Network)

```python
fnn = keras.Sequential([
    layers.Dense(16, activation="relu", input_shape=(2,)),
    layers.Dense(8, activation="relu"),
    layers.Dense(1, activation="sigmoid"),
])
fnn.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
fnn.fit(X_tr, y_tr, epochs=10, verbose=0)
```

## 3.4 CNN (on synthetic image-like data)

```python
X_img = np.random.rand(100, 8, 8, 1)
y_img = np.random.randint(0, 2, 100)

cnn = keras.Sequential([
    layers.Conv2D(8, (3, 3), activation="relu", input_shape=(8, 8, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(16, activation="relu"),
    layers.Dense(1, activation="sigmoid"),
])
cnn.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
cnn.fit(X_img, y_img, epochs=5, verbose=0)
```

## 3.5 RNN (on synthetic sequence data)

```python
X_seq = np.random.rand(100, 10, 1)
y_seq = np.random.randint(0, 2, 100)

rnn = keras.Sequential([
    layers.SimpleRNN(16, input_shape=(10, 1)),
    layers.Dense(8, activation="relu"),
    layers.Dense(1, activation="sigmoid"),
])
rnn.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
rnn.fit(X_seq, y_seq, epochs=5, verbose=0)
```

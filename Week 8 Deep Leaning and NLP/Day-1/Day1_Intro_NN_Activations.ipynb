# Day 1: Introduction to Deep Learning + Neural Networks + Activations

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1.1 Single Neuron (Perceptron)

```python
def neuron(x, w, b):
    z = np.dot(x, w) + b
    return z

x = np.array([0.5, -0.2, 0.8])
w = np.array([0.4, 0.3, -0.6])
b = 0.1
print("Neuron output:", neuron(x, w, b))
```

## 1.2 Activation Functions

```python
def step(x):
    return np.where(x >= 0, 1, 0)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def softmax(x):
    e = np.exp(x - np.max(x))
    return e / e.sum()

x_vals = np.linspace(-10, 10, 200)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, step(x_vals), label="Step")
plt.plot(x_vals, sigmoid(x_vals), label="Sigmoid")
plt.plot(x_vals, tanh(x_vals), label="Tanh")
plt.plot(x_vals, relu(x_vals), label="ReLU")
plt.plot(x_vals, leaky_relu(x_vals), label="Leaky ReLU")
plt.legend()
plt.title("Activation Functions")
plt.show()
```

```python
print("Softmax:", softmax(np.array([2.0, 1.0, 0.1])))
```

## 1.3 Forward Pass of a Tiny Neural Network (2-3-1)

```python
np.random.seed(42)
X = np.random.randn(5, 2)

W1 = np.random.randn(2, 3)
b1 = np.zeros((1, 3))
W2 = np.random.randn(3, 1)
b2 = np.zeros((1, 1))

def forward_pass(X):
    z1 = X @ W1 + b1
    a1 = relu(z1)
    z2 = a1 @ W2 + b2
    a2 = sigmoid(z2)
    return a2

print("Forward pass output:\n", forward_pass(X))
```

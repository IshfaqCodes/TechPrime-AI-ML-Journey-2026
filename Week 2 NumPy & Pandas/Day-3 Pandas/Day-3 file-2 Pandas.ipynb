### Handling Missing Values

```python
df.isnull().sum() #checking null values ager kam hay to drop warna fill
```

### Drop missing values
To drop rows with missing values

```python
df.dropna() #dropna drop the row with missing value But hum direct drop nhi krengee fill karnee ki koshish krengee
df.isnull().sum()
```

### Fill missing values
Handling missing values is an important part of Exploratory Data Analysis

```python
df["age"]=df["age"].fillna(df["age"].mean()) #hamree pass age main 7 values null hay to hum age ka mean nekal krr fill krthee hay 
df.isnull().sum() #age ka mean 53 hay to null values 53 se fill huwee hay
```

```python
#Loding new dataset
df2=pd.read_csv("sales_data.csv")
df2
```

```python

#Convert to lowercase:
df2["Product_Category"] = df2["Product_Category"].str.lower()

#Convert to uppercase:
df2["Product_Category"] = df2["Product_Category"].str.upper()

#Remove extra spaces:
df2["Product_Category"] = df2["Product_Category"].str.strip()

df2
```

# Importing Pandas

```python
import pandas as pd #pandas ko short main pd se change kia 
import numpy as np
```

## Creating a pandas series
 

```python
numbers = [10, 20, 30, 40] #list of numbers 
series = pd.Series(numbers) # creating a series 
print(series) #Yaha per hum ne print kia hay series ko
```

```python
numbers=np.array([1,2,3,4,5,6,7]) #Numpy array ka use krr k series kese banayee let see
series=pd.Series(numbers) #creating series for numpy array
print(series) #Yaha per hum ne print kia hay series ko
```

## Creating a pandas dataframe

```python
data={
    "Name":["Ishfaq khan","Kashif wazir","Majid wazir","Afaq khan","Wasif wazir","Hassan wazir"],
    "Age":[20,22,28,25,20,29],
    "Marks":[792,845,840,826,675,899]
    }
df=pd.DataFrame(data)
df.reset_index()
```

### Head Function
This shows the first 5 rows of the dataset or DataFrame

```python
df.head().reset_index()
```

### Tail Function
This shows the last 5 rows of the ataset or DataFrame

```python
df.tail().reset_index()
```

### Shape Attribute

```python
print(df) #To Print the dataframe
print("Shape=",df.shape) # To Print the shape of dataframe
```

### Columns Attribute

```python
df.columns #To check the columns 
```

## Selecting Columns from DataFrame

```python
print("\n Age Column is =\n",df["Age"]) #Select a single column
print("\n Name and Marks columns is =\n",df[["Name","Marks"]]) #select a multiple columns
print("\n Age,Marks,Name columns is =\n",df[["Age","Marks","Name"]])#select a multiple columns
```

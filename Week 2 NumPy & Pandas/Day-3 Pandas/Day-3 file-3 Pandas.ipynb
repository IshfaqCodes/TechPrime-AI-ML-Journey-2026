# Importing library

```python
import pandas as pd
```

### Read csv file 

```python
df=pd.read_csv("sales_data.csv")
df.head()
```

### GroupBy
Common aggregation functions sum(),mean(),count(),min(),max(),median()

```python
#Groupby with single column
sales_region=df.groupby("Region")["Sales_Amount"].sum()
sales_min=df.groupby("Region")["Sales_Amount"].min()
sales_max=df.groupby("Region")["Sales_Amount"].max()
sales_count=df.groupby("Region")["Sales_Amount"].count()

print("\n Region Sales:\n",sales_region)
print("\n Region Minimam Sales:\n",sales_min)
print("\n Region Maximam Sales:\n",sales_max)
print("\n Region Sales Count:\n",sales_count)


```

### Groupby with multiple columns

```python
sales_region=df.groupby(["Region","Product_Category"])["Sales_Amount"].sum()
sales_min=df.groupby(["Region","Product_Category"])["Sales_Amount"].min()
sales_max=df.groupby(["Region","Product_Category"])["Sales_Amount"].max()
sales_count=df.groupby(["Region","Product_Category"])["Sales_Amount"].count()

print("\n Region Sales:\n",sales_region)
print("\n Region Minimam Sales:\n",sales_min)
print("\n Region Maximam Sales:\n",sales_max)
print("\n Region Sales Count:\n",sales_count)
```

### Using Multiple Aggregation Functions

```python
df1=df.groupby("Region")["Sales_Amount"].agg(["max","min","sum"]) #multiple aggregation with single column
df1.reset_index()
```

```python
df2=df.groupby(["Region","Product_Category"])["Sales_Amount"].agg(["mean","max","min","sum"])#multiple aggregation with 2 columns
df2

```

### Using agg with Dictionary

```python
df1=df.groupby("Region").agg({
    "Sales_Amount":["sum","mean","min","max"]
})
df1
```

###  Resetting Index After GroupBy
This makes the output cleaner and easier to use

```python
df1=df.groupby("Region").agg({
    "Sales_Amount":["sum","mean","min","max"],
    "Quantity_Sold":["sum","mean","min","max"],
    "Unit_Cost":["sum","mean","min","max"],
    "Unit_Price":["sum","mean","min","max"]
}).reset_index()
df1
```

### Counting Values Using GroupBy

```python
df4=df.groupby("Region")["Sales_Amount"].count().reset_index()
df4
```

### Slicing

```python
numbers=np.array([1,2,3,4,5,8]) # 3 to 8
print(numbers[2:6]) #yaha per start huwa hay index 2=3 awer end number hay index 6 jo k shumar nhi hotha to os se pehla number shumar hoga index 5=8
```

```python
#Another example Reverse
numbers=np.array([1,2,3,4]) #4 to 1
print(numbers[::-1])#
```

```python
# slicing with 2D array
arr2=np.array([[22,12,33,44,55],[88,77,99,55,55]])
print(arr2[0,0:3])
print(arr2[1,0:3])
print(arr2[0,::-1])
print(arr2[1,::-1])

```

### 3D array
